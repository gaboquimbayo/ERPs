## NEURA DATA

```python
# path MA files
P3_MA_path = "ERPs/ERP_Criterion_Good/Measures/P3_MA.txt"
N2_MA_path = "ERPs/ERP_Criterion_Good/Measures/N2_MA.txt"

# path Latency files
P3_Latency_path = "ERPs/ERP_Criterion_Good/Measures/P3_Latency.txt"
N2_Latency_path = "ERPs/ERP_Criterion_Good/Measures/N2_Latency.txt"

# read the MA files
p3_ma = pd.read_csv(P3_MA_path, delim_whitespace=True)
n2_ma = pd.read_csv(N2_MA_path, delim_whitespace=True)

# read the Latency files
p3_lat = pd.read_csv(P3_Latency_path, delim_whitespace=True)
n2_lat = pd.read_csv(N2_Latency_path, delim_whitespace=True)

# Extract the subject number from the MA file name
p3_ma['Subject'] = p3_ma['ERPset'].str.extract('(\d+)$').astype(int)
n2_ma['Subject'] = n2_ma['ERPset'].str.extract('(\d+)$').astype(int)

# Extract the subject number from the Latency file name
p3_lat['Subject'] = p3_lat['ERPset'].str.extract('(\d+)$').astype(int)
n2_lat['Subject'] = n2_lat['ERPset'].str.extract('(\d+)$').astype(int)

# Set the subject number as index - MA
p3_ma.set_index('Subject', inplace=True)
n2_ma.set_index('Subject', inplace=True)

# Set the subject number as index - Latency
p3_lat.set_index('Subject', inplace=True)
n2_lat.set_index('Subject', inplace=True)

# drop the ERPset column - MA
p3_ma.drop(columns=['ERPset'], inplace=True)
n2_ma.drop(columns=['ERPset'], inplace=True)

# drop the ERPset column - Latency
p3_lat.drop(columns=['ERPset'], inplace=True)
n2_lat.drop(columns=['ERPset'], inplace=True)

# drop the bin2 columns - MA
p3_ma = p3_ma.loc[:, ~p3_ma.columns.str.contains('bin2')]
n2_ma = n2_ma.loc[:, ~n2_ma.columns.str.contains('bin2')]

# drop the bin2 columns - Latency
p3_lat = p3_lat.loc[:, ~p3_lat.columns.str.contains('bin2')]
n2_lat = n2_lat.loc[:, ~n2_lat.columns.str.contains('bin2')]

# change the column names - MA
p3_ma.columns = [col.replace('bin1', 'P3_B1') for col in p3_ma.columns]
p3_ma.columns = [col.replace('bin3', 'P3_B1B2') for col in p3_ma.columns]
n2_ma.columns = [col.replace('bin1', 'N2_B1') for col in n2_ma.columns]
n2_ma.columns = [col.replace('bin3', 'N2_B1B2') for col in n2_ma.columns]

# change the column names - Latency
p3_lat.columns = [col.replace('bin1', 'P3_B1') for col in p3_lat.columns]
p3_lat.columns = [col.replace('bin3', 'P3_B1B2') for col in p3_lat.columns]
n2_lat.columns = [col.replace('bin1', 'N2_B1') for col in n2_lat.columns]
n2_lat.columns = [col.replace('bin3', 'N2_B1B2') for col in n2_lat.columns]

# add the suffix to the column names - MA
p3_ma.columns = [col + '_MA' for col in p3_ma.columns]
n2_ma.columns = [col + '_MA' for col in n2_ma.columns]

# add the suffix to the column names - Latency
p3_lat.columns = [col + '_LAT' for col in p3_lat.columns]
n2_lat.columns = [col + '_LAT' for col in n2_lat.columns]
```

```python
data = data.merge(p3_ma, left_on='Subject', right_index=True)
data = data.merge(n2_ma, left_on='Subject', right_index=True)
data = data.merge(p3_lat, left_on='Subject', right_index=True)
data = data.merge(n2_lat, left_on='Subject', right_index=True)
```

```python
# read data of the second file. this file contains the "Neural correlate" variable
bands_path = 'CRB_Power_B1B2.xlsx'
bands_data = pd.read_excel(bands_path)
bands_data = bands_data.rename(columns={'subid': 'Subject'})
bands_data = bands_data.drop(columns=['Freq_Range'])
```

```python
for bin in ['B1', 'B1B2']:
    for band in bands_data['Freq_Names'].unique():
        neural_data = bands_data[bands_data['Freq_Names'] == band][['Subject'] + [f'{electrode}_{bin}' for electrode in electrodos_eeg]]
        neural_data = neural_data.rename(columns={f'{electrode}_{bin}': f'{band}_{electrode}_{bin}' for electrode in electrodos_eeg})
        neural_data.set_index('Subject', inplace=True)
        data = data.merge(neural_data, left_on='Subject', right_index=True)
```
